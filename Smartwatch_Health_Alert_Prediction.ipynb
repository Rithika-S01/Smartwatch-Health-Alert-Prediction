```python
import pandas as  pd
import numpy as np
```

```python
df =pd.read_csv("smartwatch_health_alerts_datasets - LMS.csv")
df
```

```python
print(df.info())
print(df.describe())
```

```python
df.isnull().sum()
```

```python
import numpy as np
N_col=df.select_dtypes(include=np.number)

for i in N_col:
    df[i]=df[i].fillna(df[i].median())
```

```python
df.isnull().sum()
```

```python
df['workout_type']=df['workout_type'].fillna(df['workout_type'].mode()).iloc[0]
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='health_alert_type', data=df)
```

```python
df1 = df.select_dtypes(include=['number'])
plt.figure(figsize=(12, 6))
sns.heatmap(df1.corr(),annot=True,fmt=".1f")
```

```python
df['health_alert_type'].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler,LabelEncoder
for col in df.select_dtypes(include=['object']).columns:
    l=LabelEncoder()#calling function
    l.fit(df[col])# is to fit the data to the function
    df[col]=l.transform(df[col])# transforming to numberical data
df
```

```python
from sklearn.model_selection import train_test_split 

#splitting the dataset into training and testing
x=df.drop('health_alert_type',axis=1)
y=df['health_alert_type']

#Standardize the features using standardscaler
scaler=StandardScaler()
x=scaler.fit_transform(x)
from imblearn.over_sampling import SMOTE #synthetic minority oversampling technique

#apply 2: apply smote to balance the classes in the training set 
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
smote=SMOTE()

x_train_res,y_train_res=smote.fit_resample(x_train,y_train)
```

```python
from sklearn.linear_model import LogisticRegression
logreg=LogisticRegression()
logreg.fit(x_train_res,y_train_res)
```

```python
from sklearn.metrics import accuracy_score,roc_auc_score,confusion_matrix,classification_report
#step 4:Model evalution
y_pred=logreg.predict(x_test)
accuracy=accuracy_score(y_test,y_pred)
conf=confusion_matrix(y_test,y_pred)
clas= classification_report(y_test,y_pred)


print("accruracy",accuracy)
print("confusion matrix",conf)
print("classification report\n",clas)
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc =DecisionTreeClassifier()
dtc.fit(x_train_res,y_train_res)
```

```python
from sklearn.metrics import accuracy_score,roc_auc_score,confusion_matrix,classification_report
#step 4:Model evalution
y_pred_1=dtc.predict(x_test)
accuracy_dtc=accuracy_score(y_test,y_pred_1)
conf_dtc=confusion_matrix(y_test,y_pred_1)
clas_dtc= classification_report(y_test,y_pred_1)


print("accruracy",accuracy_dtc)
print("confusion matrix",conf_dtc)
print("classification report\n",clas_dtc)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Step 1: Initialize model
rfc = RandomForestClassifier(random_state=42)

# Step 2: Train model
rfc.fit(x_train_res, y_train_res)

# Step 3: Prediction
y_pred_rf = rfc.predict(x_test)

# Step 4: Model Evaluation
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix, classification_report

accuracy_rf = accuracy_score(y_test, y_pred_rf)
conf_rf = confusion_matrix(y_test, y_pred_rf)
clas_rf = classification_report(y_test, y_pred_rf)

print("Accuracy:", accuracy_rf)
print("Confusion Matrix:\n", conf_rf)
print("Classification Report:\n", clas_rf)
```
